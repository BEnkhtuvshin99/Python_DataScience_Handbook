```python
import pandas as pd 
df=pd.read_csv(f"data/president_heights.csv")
```

```python
heights = np.array(df['height(cm)'])
```

```python
heights
```

```python
print(f"Mean heights:",heights.mean())
print(f"Standard deviation:",heights.std())
print(f"Minimum heights:",heights.min())
print(f"Maximumheights:",heights.max())
```

```python
print(f"25th percentile:",np.percentile(heights,25))
print(f"median:",np.median(heights))
print(f"75th percentile:",np.percentile(heights,75))
```

```python
%matplotlib inline 
import matplotlib.pyplot as plt
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
```

```python
plt.hist(heights)
plt.title("Presidents height distribution")
plt.xlabel('height(cm)')
plt.ylabel('number')
```
