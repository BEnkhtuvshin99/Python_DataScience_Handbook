```python
rng=np.random.default_rng(seed=1701)
```

```python
def calculate(values):
    output=np.empty(len(values))
    for i in range(len(values)):
        output[i]=1/values[i]
    return output
```

```python
print(rng.integers(1,10,5))
```

```python
values=rng.integers(1,10,5)
calculate(values)
```

```python
values=rng.integers(1,100,size=10000)
```

```python
%timeit calculate(values)
```
