```python
import yfinance as yf

sp500 = yf.download('^GSPC', start='2018-01-01', end='2022-12-31')
print(sp500.head())
```

```python
sp500.head()
```

```python
sp500=sp500['Close']
```

```python
%matplotlib inline 
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
sp500.plot()
```

```python
sp500.plot(alpha=0.5,style='-')
sp500.resample('BYE').mean().plot(style=':')
sp500.asfreq('BYE').plot(style='--');
plt.legend(['input','resample','asfreq'],loc='upper left')
```

```python
fig, ax =plt.subplots(2,sharex=True)
data=sp500.iloc[:20]
data.asfreq('D').plot(ax=ax[0],marker='o')
data.asfreq('D',method='bfill').plot(ax=ax[1],style='-o')
data.asfreq('D',method='ffill').plot(ax=ax[1],style='--o')
ax[1].legend(["back-fill","forward=fill"]);
```

```python
sp500=sp500.asfreq('D',method='pad')
```

```python
ROI =100*(sp500.shift(-365)-sp500)/sp500
ROI.plot()
plt.ylabel('% Return on Investment after 1 year')
```

```python
rolling = sp500.rolling(365,center=True)
```

```python
rolling
```

```python
data=pd.DataFrame({'input':sp500,'one-year rolling_mean':rolling.mean(),'one-year rolling_median':rolling.median()})
ax=data.plot(style=['-','--',':'])
ax.lines[0].set_alpha(0.3)
```

```python
data=pd.read_csv(r"C:\Users\benkh\Python_Handbook\Handbook\data\FremontBridge.csv",index_col='Date',parse_dates=True)
```

```python
data.head()
```

```python
data.columns=['Total','East','West']
```

```python
data.dropna().describe()
```

```python
weekly=data.resample('W').sum()
weekly.plot(style=['-',':','--'])
plt.ylabel('Weekly bicycle count')
```

```python
daily = data.resample('D').sum()
daily.rolling(30,center=True).sum().plot(style=['-',':','--'])
plt.ylabel('mean hourly count')
```

```python
daily.rolling(50,center=True,win_type='gaussian').sum(std=10).plot(style=['-',':','--'])
```

```python
by_time = data.groupby(data.index.time).mean()
hourly_ticks = 4*60*60*np.arange(6)
by_time.plot(xticks=hourly_ticks,style=['-',':','--'])
```

```python
by_weekday=data.groupby(data.index.dayofweek).mean()
by_weekday.index=['Mon','Tues','Wed','Thurs','Fri','Sat','Sun']
by_weekday.plot(style=['-',';','--']);
```

```python
weekend =np.where(data.index.weekday<5,'Weekday','Weekend')
by_time=data.groupby([weekend,data.index.time]).mean()
```

```python
import matplotlib.pyplot as plt
fig, ax =plt.subplots(1,2,figsize=(14,15))
by_time.loc['Weekday'].plot(ax=ax[0],title='Weekdays',xticks=hourly_ticks,style=['-',':','--'])
by_time.loc['Weekend'].plot(ax=ax[1],title='Weekends',xticks=hourly_ticks,style=['-',':','--']);
```
