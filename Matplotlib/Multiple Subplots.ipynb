```python
mean = [0, 0]
cov = [[1, 1], [1, 2]]
rng = np.random.default_rng(1701)
x, y = rng.multivariate_normal(mean, cov, 3000).T

# Create figure and grid
fig = plt.figure(figsize=(6, 6))
grid = plt.GridSpec(4, 4, hspace=0.2, wspace=0.2)

main_ax = fig.add_subplot(grid[:-1, 1:])
y_hist = fig.add_subplot(grid[:-1, 0], sharey=main_ax)
x_hist = fig.add_subplot(grid[-1, 1:], sharex=main_ax)

# Scatter plot
main_ax.plot(x, y, 'ok', markersize=3, alpha=0.2)

# Histograms
x_hist.hist(x, bins=40, histtype='stepfilled', orientation='vertical')
x_hist.invert_yaxis()

y_hist.hist(y, bins=40, histtype='stepfilled', orientation='horizontal')
y_hist.invert_xaxis()

# Clean up ticks (this is important)
y_hist.tick_params(axis="y", labelleft=False)
x_hist.tick_params(axis="x", labelbottom=False)

# Optional labels
main_ax.set_xlabel("X")
main_ax.set_ylabel("Y")

plt.show()
```
